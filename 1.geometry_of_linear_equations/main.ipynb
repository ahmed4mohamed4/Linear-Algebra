# Geometry of Linear Equations

```python
# Importing
import numpy as np
import matplotlib.pyplot as plt
```

## Row picture

```python
# x values for drawing the lines
x = np.linspace (-2, 4, 200)

# Equations
y1 = 2 * x # 2x_1 - x_2 = 0$
y2 = (x + 3) / 2 # -x_1 + 2x_2 = 3$

# Plot the lines
plt.figure (figsize=(6, 6))
plt.plot (x, y1, label= r"$2x_1 - x_2 = 0$")
plt.plot (x, y2, label= r"$-x_1 + 2x_2 = 3$")

# Plot the solution
plt.scatter (1, 2, s=80)
plt.text (1.2, 2.2, "(1, 2)")

# Axes
plt.axhline (0) # x
plt.axvline (0) # y

plt.xlim (-2, 4)
plt.ylim (-2, 4)

plt.xlabel (r"$x_1$")
plt.ylabel (r"$x_2$")
plt.title ("Row Picture")
plt.grid (True)
plt.legend ()

plt.show()
```

## Column Picture

```python
# Matrix A (Coefficients)
A = np.array ([[2, -1],
               [-1, 2]])

# Vector b (right hand side)
b = np.array ([0, 3])

# Solve Ax = b
x = np.linalg.solve (A, b)

# Columns of A
c1 = A [:, 0] # First column
c2 = A [:, 1] # Second column

# Scaled columns
v1 = x [0] * c1
v2 = x [1] * c2

plt.figure (figsize= (6, 6))

# Axes
plt.axhline (0, color="black")
plt.axvline (0, color="black")

# Original columns
plt.quiver (0, 0, c1 [0], c1 [1],
           angles= 'xy', scale_units= 'xy', scale= 1,
           color= 'blue', label= 'Column 1')

plt.quiver (0, 0, c2 [0], c2 [1],
           angles= 'xy', scale_units= 'xy', scale= 1,
           color= 'green', label= 'Column 2')

# x1 * Column1
plt.quiver (0, 0, v1 [0], v1 [1],
           angles= 'xy', scale_units= 'xy', scale= 1,
           color= 'red', label= 'x1 * Column1')

# x2 * Column2 (starts from the end of v1)
plt.quiver (v1 [0], v1 [1], v2 [0], v2 [1],
           angles= 'xy', scale_units= 'xy', scale= 1,
           color= 'orange', label= 'x2 * Column2')

# Result vector b
plt.quiver (0, 0, b [0], b [1],
           angles= 'xy', scale_units= 'xy', scale= 1,
           color= 'purple', linewidth= 3, label= 'b')

plt.text (b [0] + 0.1, b [1], "b")

plt.xlim (-3, 3)
plt.ylim (-3, 5)
plt.gca ().set_aspect ('equal')

plt.grid (True)
plt.legend ()
plt.title ("Column Picture")

plt.show ()
```

```python
# Matrix A (Coefficients)
A = np.array ([[2, -1],
               [-1, 2]])
```

```python
# Vector b (right hand side)
b = np.array ([0, 3])
```

```python
x = np.linalg.solve (a= A, b= b)
# Solve both equations:
print (f"(x, y) = ({x [0]}, {x[1]})")
```

```python
# To make sure
# This will return b [0, 3]
vector_b = (A @ x) # Matrix Multiplication
print (f"b = [{vector_b [0]}, {vector_b [1]}]")
```
